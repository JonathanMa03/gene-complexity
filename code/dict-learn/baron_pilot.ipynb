# Baron dictionary-learning pilot

Use the project kernel and Run All. Requires `baron_classical_v1`. Fits only training cells; reports validation reconstruction and sparsity. Never reads test features or fits classifiers. Defaults are a small provisional pilot, not tuned choices. Each run preserves earlier outputs.


```python
from pathlib import Path
from uuid import uuid4
import sys
import pandas as pd
import matplotlib.pyplot as plt
root = next(p for p in (Path.cwd(), *Path.cwd().parents) if (p / "code/pipeline/dictionary_pilot.py").exists())
sys.path.insert(0, str(root / "code/pipeline"))
from dictionary_pilot import run
output = root / "results/dict-learning" / f"baron_notebook_{uuid4().hex[:8]}"
report = run(root / "data/processed/baron_classical_v1", output, atoms=16, alpha=1.0, epochs=10)
```

```python
metrics = pd.DataFrame.from_dict(report["metrics"], orient="index")
display(metrics)
print("Reached epoch budget:", report["reached_epoch_budget"])
print("Solver warnings:", report["warnings"])
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
metrics[["mse", "zero_reconstruction_mse"]].plot.bar(ax=axes[0], rot=0)
metrics["active_fraction"].plot.bar(ax=axes[1], rot=0, ylim=(0, 1))
axes[0].set_ylabel("Reconstruction MSE")
axes[1].set_ylabel("Fraction of coefficients active")
plt.tight_layout()
plt.show()
print("Artifacts:", output)
```

## Interpretation

Check warnings and the epoch budget before treating the fit as settled. Reconstruction is not a classification metric. Codes are not expected to have the same dimension as scGPT embeddings. Use cell IDs for alignment; do not select hyperparameters based on final test performance.
